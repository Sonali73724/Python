```python
import pandas as pd

df = pd.read_csv("tesla_deliveries_dataset_2015_2025.csv")
df.head()
```

```python
df.shape
df.info()
df.describe()
df.columns
```

```python
df.isnull().sum()
df = df.dropna()
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
import matplotlib.pyplot as plt

df_year = df.groupby("Year")["Estimated_Deliveries"].sum()

plt.plot(df_year.index, df_year.values)
plt.title("Tesla Deliveries Trend (2015-2025)")
plt.show()
```

```python
df.columns
```

```python
df_year2 = df.groupby("Year")["Production_Units"].sum()

plt.plot(df_year2.index, df_year2.values)
plt.title("Tesla Production Trend")
plt.show()
```

```python
plt.plot(df_year.index, df_year.values, label="Deliveries")
plt.plot(df_year2.index, df_year2.values, label="Production")
plt.legend()
plt.title("Production vs Deliveries")
plt.show()
```

```python
region = df.groupby("Region")["Estimated_Deliveries"].sum().sort_values()
region.plot( kind="bar")
plt.title("Region-wise Deliveries")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.plot(df["Year"], df["Estimated_Deliveries"])
plt.title("Tesla Deliveries Over Years")
plt.xlabel("Year")
plt.ylabel("Deliveries")
plt.show()
```

```python
plt.plot(df["Year"],df["Production_Units"])
plt.title("Tesla Production Units Over Years")
plt.xlabel("Year")
plt.ylabel("Production Units")
plt.show()
```

```python
plt.plot(df["Year"],df["Production_Units"],label="Production")
plt.plot(df["Year"],df["Estimated_Deliveries"],label="Deliveries")
plt.title("Tesla Production and Deliveries Over Years")
plt.xlabel("Year")
plt.ylabel("Units")
plt.legend()
```
