# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
# 🔧 TOOL 3: Word Counter

def word_counter(text: str):
    """Count words in a sentence."""
    return len(text.split())
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (TO IMPLEMENT)

def agent(query: str):

    query_lower = query.lower()

    print("="*40)
    print("User Query:", query)

    # Calculator Routing
    if "calculate" in query_lower:

        expression = query_lower.replace("calculate", "").strip()

        result = calculator(expression)

        print("Intent Detected : Calculation")
        print("Tool Used       : Calculator")

        return {
            "type": "calculation",
            "result": result
        }

    # Keyword Routing
    elif "keyword" in query_lower or "extract" in query_lower:

        text = query.replace("Extract keywords from", "").strip()

        result = extract_keywords(text)

        print("Intent Detected : Keyword Extraction")
        print("Tool Used       : Keyword Extractor")

        return {
            "type": "keywords",
            "result": result
        }

    # Word Counter (Bonus Tool)
    elif "count words" in query_lower:

        text = query.replace("count words", "").strip()

        result = word_counter(text)

        print("Intent Detected : Word Count")
        print("Tool Used       : Word Counter")

        return {
            "type": "word_count",
            "result": result
        }

    # General Query
    else:

        print("Intent Detected : General")
        print("Tool Used       : Direct Response")

        return {
            "type": "general",
            "result": "This is a general query."
        }
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
     "Count words Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")
    if user_input.lower() == "exit":
        break
    print("Response:", agent(user_input))
````
